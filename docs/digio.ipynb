```python
import base64

def pdf_to_base64(pdf_path: str) -> str:
    """
    Reads a PDF file and converts it to a Base64 encoded string.
    
    :param pdf_path: Path to the PDF file
    :return: Base64 encoded string
    """
    with open(pdf_path, "rb") as pdf_file:
        encoded_string = base64.b64encode(pdf_file.read())
        return encoded_string.decode("utf-8")

base64_pdf = pdf_to_base64("/Users/anshagarwal/Desktop/KirayaEase/data/residential-rental-agreement-format.pdf")
```

```python
import os
from dotenv import load_dotenv
load_dotenv()
DIGIO_CLIENT_ID = os.getenv("DIGIO_CLIENT_ID")
DIGIO_CLIENT_SECRET = os.getenv("DIGIO_CLIENT_SECRET")
```

```python
def _basic_auth_header(client_id: str, client_secret: str) -> str:
    token = base64.b64encode(f"{client_id}:{client_secret}".encode()).decode()
    return f"Basic {token}"
```

```python
import requests
headers = {
        "Authorization": _basic_auth_header(DIGIO_CLIENT_ID, DIGIO_CLIENT_SECRET),
        "Content-Type": "application/json",
    }
try:
    resp = requests.post(DIGIO_CREATE_SIGN_URL, headers=headers, json=body, timeout=30)
except requests.RequestException as e:
    raise HTTPException(status_code=502, detail=f"Network error calling Digio: {e}")
```
