from flask import Flask, request, jsonify import requests

app = Flask(**name**)

###Authorisation credentials CLIENT_ID = “YOUR_DIGILOCKER_CLIENT_ID”
CLIENT_SECRET = “YOUR_DIGILOCKER_CLIENT_SECRET” REDIRECT_URI =
“YOUR_REGISTERED_REDIRECT_URI” TOKEN_URL =
“https://digilocker.meripehchaan.gov.in/public/oauth2/1/token”

@app.route(‘/digilocker/auth’, methods=\[‘POST’\]) def
digilocker_auth(): data = request.json auth_code =
data.get(‘auth_code’)  
##Auth code received after user authorizes

    ###Exchange code for access token
    payload = {
        'grant_type': 'authorization_code',
        'code': auth_code,
        'client_id': CLIENT_ID,
        'client_secret': CLIENT_SECRET,
        'redirect_uri': REDIRECT_URI,
    }
    headers = {'Content-Type': 'application/x-www-form-urlencoded'}
    response = requests.post(TOKEN_URL, data=payload, headers=headers)

    if response.status_code == 200:
        token_data = response.json()
        access_token = token_data.get('access_token')
        # Here you could fetch user info/documents if needed
        print("User authenticated successfully!")
        return jsonify({"auth_success": True, "access_token": access_token})
    else:
        print("Authentication failed:", response.text)
        return jsonify({"auth_success": False, "error": response.text}), 400

if **name** == ‘**main(variable name)**’: app.run(port=5000, debug=True)
